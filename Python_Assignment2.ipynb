## Data Structures – List, Dictionary, Set & Conditional Statements



## 1. List Creation

```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Binu", "Charu", "Divya", "Esha"]

print("Age list:", age_list)
print("Name list:", name_list)
```

## 2. List Operations / Modifications

```python
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After inserting 30:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After removing Yazhini:", name_list)

# d. Pop the last element from age_list
removed_age = age_list.pop()
print("Popped age:", removed_age)
print("After pop:", age_list)

# e. Extend age_list with additional ages
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Find maximum, minimum, and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

## 3. Accessing List Elements


```python
print("First element:", name_list[0])
print("Last element:", name_list[-1])
print("Elements from index 2 to index 4:", name_list[2:5])
print("Names in reverse order:", name_list[::-1])
```

## 4. Dictionary – Creation, Modification and Access

```python

student_marks = {
    "Anu": 91,
    "Binu": 78,
    "Charu": 85,
    "Divya": 88,
    "Esha": 95
}
print("Student marks:", student_marks)

# b. Access and print one student's mark
print("Anu's mark:", student_marks["Anu"])

# c. Add Janani with a mark of 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update an existing student's mark to 82
student_marks["Binu"] = 82
print("After updating Binu's mark:", student_marks)

# e. Print all keys, values, and key-value pairs
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

## 5. Sets – Operations

```python
# a. Create a set 
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print("my_set:", my_set)
# Sets keep only unique elements. Repeated 'a' and 'i' appear once. Sets are unordered, so display order may vary.

# b. Attempt to change an item by index
try:
    my_set[4] = 's'
except TypeError as error:
    print("Error:", error)
  # Sets are unordered and do not support indexing or item assignment.

# c. Create set1 and set2
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d. Compute and print union and intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

## 6. Operators & Conditional Statements – Performance Category

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent work! Keep it up.")
elif score >= 4:
    print("Average: Good effort! Keep practicing, there is room for improvement.")
else:
    print("Below Average: Need to improve your performance; consistent practice will lead to better results.")
```
