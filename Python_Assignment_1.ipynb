# Python Assignment 1: Data Structures – Strings & Tuples


## 1. String Concatenation


```python

str1 = "Hello "
name = input("Enter your Name: ")
str2 = name

result = str1 + str2
print(result)

str3 = ", welcome to Python programming"
result = result + str3
print(result)
```

## 2. String Slicing and Indexing

```python
name = input("Enter your Name: ")
text = "Hello " + name + ", welcome to Python programming"

print("First character:", text[0])
print("Last character:", text[-1])
print("First 5 characters:", text[:5])
print("Last 11 characters:", text[-11:])
print("Reverse:", text[::-1])
print("Word Python:", text[25:31])
```

## 3. String Methods


```python
strM = "Python beginner tutorial"

print("Uppercase:", strM.upper())
print("Lowercase:", strM.lower())
print("Capitalized:", strM.capitalize())
print("Number of occurrences of t:", strM.count("t"))
print("After replacement:", strM.replace("Python", "Machine Learning"))
```

## Tuples – Creation, Modification and Access

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t_combine = t1 + t2
print("Combined tuple:", t_combine)

t_repeat = t_combine * 3
print("Tuple repeated 3 times:", t_repeat)

print("3rd element:", t_combine[2])
print("First three elements:", t_combine[:3])
print("Last three elements:", t_combine[-3:])
```
